### 1. The Consumer's Best Response (Follower / Lower-Level Problem $\mathcal{P}_i(x)$)

The daily problem of each community member $i \in \mathcal{I}$ (the follower) is to minimize their total electricity cost and disutility over the day in response to the hourly internal price vector $x = (x_t)_{t \in \mathcal{T}}$ posted by the community manager (the leader). Inside the community, this uniform price $x_t$ replaces the day-ahead energy price and grid tariffs that the member would face alone: both net imports and net exports behind the shared meter are settled at the exact same price $x_t$.

The input data for member $i$ is: $\mathcal{T} = \{0, \dots, 23\}$, the hourly time steps; $x_t$, the internal price posted by the manager in hour $t$ (treated as a fixed parameter by the follower); $c_i^L$, the linear disutility penalty coefficient; $\ell_{i,t}^{ref} = L^{\max} \rho_t$, the preferred hourly consumption reference profile; $E_{i,min} = \sum_{t=0}^{23} \ell_{i,t}^{ref}$, the minimum daily energy requirement; $c_i^{PV}$, the marginal generation cost of the PV panels; $PV_{i,t}^{max}$, the forecasted available PV energy in hour $t$; $L_{i,t}^{min}, L_{i,t}^{max}$, the minimum and maximum hourly operational bounds of the load; and, for battery owners ($i \in \mathcal{I}^{bat}$), $\overline{E}_i, \overline{P}_i^{ch}, \overline{P}_i^{dis}, \eta_i^{ch}, \eta_i^{dis}, E_{i,0}$.

Because buying and selling inside the community occur at the same price $x_t$, the separate gross import and export variables $q_{i,t}^{imp}, q_{i,t}^{exp} \ge 0$ merge into a single unrestricted net consumption variable $q_{i,t} = q_{i,t}^{imp} - q_{i,t}^{exp} \in \mathbb{R}$. The decision variables of member $i$ are therefore $y_i = (L_{i,t}, PV_{i,t}, q_{i,t}, \Delta_{i,t}^+, \Delta_{i,t}^-, p_{i,t}^{ch}, p_{i,t}^{dis}, e_{i,t})_{t \in \mathcal{T}}$ (with $p_{i,t}^{ch} = p_{i,t}^{dis} = e_{i,t} = 0$ if $i \notin \mathcal{I}^{bat}$).

The lower-level problem $\mathcal{P}_i(x)$ for member $i \in \mathcal{I}$ is:

$$\min_{\substack{L_{i,t}, PV_{i,t}, q_{i,t}, \Delta_{i,t}^+, \Delta_{i,t}^-, \\ p_{i,t}^{ch}, p_{i,t}^{dis}, e_{i,t}}} \quad \sum_{t=0}^{23} \left( x_t q_{i,t} + c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right)$$

$$\begin{aligned} \text{s.t.} \quad L_{i,t} - PV_{i,t} + p_{i,t}^{ch} - p_{i,t}^{dis} - q_{i,t} &= 0 && : \lambda_{i,t} && \forall t \in \mathcal{T} \\ L_{i,t} - \ell_{i,t}^{ref} - \Delta_{i,t}^+ + \Delta_{i,t}^- &= 0 && : \mu_{i,t}^{\Delta} && \forall t \in \mathcal{T} \\ L_{i,t}^{min} - L_{i,t} &\le 0 && : \mu_{i,t}^{-} && \forall t \in \mathcal{T} \\ L_{i,t} - L_{i,t}^{max} &\le 0 && : \mu_{i,t}^{+} && \forall t \in \mathcal{T} \\ E_{i,min} - \sum_{t=0}^{23} L_{i,t} &\le 0 && : \mu_{i,E} \\ -PV_{i,t} &\le 0 && : \nu_{i,t}^{-} && \forall t \in \mathcal{T} \\ PV_{i,t} - PV_{i,t}^{max} &\le 0 && : \nu_{i,t}^{+} && \forall t \in \mathcal{T} \\ -\Delta_{i,t}^+ &\le 0 && : \omega_{i,t}^{+} && \forall t \in \mathcal{T} \\ -\Delta_{i,t}^- &\le 0 && : \omega_{i,t}^{-} && \forall t \in \mathcal{T} \end{aligned}$$

And for members owning a battery ($i \in \mathcal{I}^{bat}$):

$$\begin{aligned} e_{i,0} - E_{i,0} - \eta_i^{ch} p_{i,0}^{ch} + \frac{1}{\eta_i^{dis}} p_{i,0}^{dis} &= 0 && : \xi_{i,0} \\ e_{i,t} - e_{i,t-1} - \eta_i^{ch} p_{i,t}^{ch} + \frac{1}{\eta_i^{dis}} p_{i,t}^{dis} &= 0 && : \xi_{i,t} && \forall t \in \{1, \dots, 23\} \\ e_{i,23} - E_{i,0} &= 0 && : \xi_{i,end} \\ -p_{i,t}^{ch} \le 0 \quad : \underline{\phi}_{i,t}^{ch}, \qquad p_{i,t}^{ch} - \overline{P}_i^{ch} &\le 0 && : \overline{\phi}_{i,t}^{ch} && \forall t \in \mathcal{T} \\ -p_{i,t}^{dis} \le 0 \quad : \underline{\phi}_{i,t}^{dis}, \qquad p_{i,t}^{dis} - \overline{P}_i^{dis} &\le 0 && : \overline{\phi}_{i,t}^{dis} && \forall t \in \mathcal{T} \\ -e_{i,t} \le 0 \quad : \underline{\psi}_{i,t}, \qquad e_{i,t} - \overline{E}_i &\le 0 && : \overline{\psi}_{i,t} && \forall t \in \mathcal{T} \end{aligned}$$

#### KKT Optimality Conditions of Member $i$

Because $\mathcal{P}_i(x)$ is a linear program with affine constraints, Slater's condition holds, strong duality is guaranteed, and the KKT conditions are both necessary and sufficient for optimality. Forming the Lagrangian $\mathcal{L}_i$ as in Question 1.b and 3.b of Assignment 1 gives the four groups of KKT conditions for each member $i \in \mathcal{I}$:

**1. Stationarity Conditions ($\forall t \in \mathcal{T}$):**


$$\begin{aligned} \frac{\partial \mathcal{L}_i}{\partial q_{i,t}} &= x_t - \lambda_{i,t} = 0 \implies \lambda_{i,t} = x_t \\ \frac{\partial \mathcal{L}_i}{\partial L_{i,t}} &= \lambda_{i,t} + \mu_{i,t}^{\Delta} - \mu_{i,t}^{-} + \mu_{i,t}^{+} - \mu_{i,E} = 0 \\ \frac{\partial \mathcal{L}_i}{\partial PV_{i,t}} &= c_i^{PV} - \lambda_{i,t} - \nu_{i,t}^{-} + \nu_{i,t}^{+} = 0 \\ \frac{\partial \mathcal{L}_i}{\partial \Delta_{i,t}^+} &= c_i^L - \mu_{i,t}^{\Delta} - \omega_{i,t}^{+} = 0 \\ \frac{\partial \mathcal{L}_i}{\partial \Delta_{i,t}^-} &= c_i^L + \mu_{i,t}^{\Delta} - \omega_{i,t}^{-} = 0 \end{aligned}$$


And for $i \in \mathcal{I}^{bat}$:


$$\begin{aligned} \frac{\partial \mathcal{L}_i}{\partial p_{i,t}^{ch}} &= \lambda_{i,t} - \eta_i^{ch} \xi_{i,t} - \underline{\phi}_{i,t}^{ch} + \overline{\phi}_{i,t}^{ch} = 0 && \forall t \in \mathcal{T} \\ \frac{\partial \mathcal{L}_i}{\partial p_{i,t}^{dis}} &= -\lambda_{i,t} + \frac{1}{\eta_i^{dis}} \xi_{i,t} - \underline{\phi}_{i,t}^{dis} + \overline{\phi}_{i,t}^{dis} = 0 && \forall t \in \mathcal{T} \\ \frac{\partial \mathcal{L}_i}{\partial e_{i,t}} &= \xi_{i,t} - \xi_{i,t+1} - \underline{\psi}_{i,t} + \overline{\psi}_{i,t} = 0 && \forall t \in \{0, \dots, 22\} \\ \frac{\partial \mathcal{L}_i}{\partial e_{i,23}} &= \xi_{i,23} + \xi_{i,end} - \underline{\psi}_{i,23} + \overline{\psi}_{i,23} = 0 \end{aligned}$$


*(Notice that $\lambda_{i,t} = x_t$ allows substituting $x_t$ directly for $\lambda_{i,t}$ in all stationarity equations, and that $\omega_{i,t}^+, \omega_{i,t}^- \ge 0$ implies $-c_i^L \le \mu_{i,t}^{\Delta} \le c_i^L$.)*

**2. Primal Feasibility, Dual Feasibility, and Complementarity Conditions ($\forall t \in \mathcal{T}$):**
Using the perpendicularity notation $0 \le a \perp b \ge 0$ from the Week 7 lecture slides (which simultaneously encodes primal inequality feasibility $a \ge 0$, dual feasibility $b \ge 0$, and complementary slackness $a \cdot b = 0$):


$$\begin{aligned} L_{i,t} - PV_{i,t} + p_{i,t}^{ch} - p_{i,t}^{dis} - q_{i,t} &= 0, \qquad L_{i,t} - \ell_{i,t}^{ref} - \Delta_{i,t}^+ + \Delta_{i,t}^- = 0 \\ 0 \le L_{i,t} - L_{i,t}^{min} &\perp \mu_{i,t}^{-} \ge 0, \qquad 0 \le L_{i,t}^{max} - L_{i,t} \perp \mu_{i,t}^{+} \ge 0 \\ 0 \le \sum_{t=0}^{23} L_{i,t} - E_{i,min} &\perp \mu_{i,E} \ge 0 \\ 0 \le PV_{i,t} &\perp \nu_{i,t}^{-} \ge 0, \qquad 0 \le PV_{i,t}^{max} - PV_{i,t} \perp \nu_{i,t}^{+} \ge 0 \\ 0 \le \Delta_{i,t}^+ &\perp \omega_{i,t}^{+} \ge 0, \qquad 0 \le \Delta_{i,t}^- \perp \omega_{i,t}^{-} \ge 0 \end{aligned}$$


And for $i \in \mathcal{I}^{bat}$:


$$\begin{aligned} e_{i,t} - e_{i,t-1} - \eta_i^{ch} p_{i,t}^{ch} + \frac{1}{\eta_i^{dis}} p_{i,t}^{dis} &= 0 \quad (\text{with } e_{i,-1} = E_{i,0}), \qquad e_{i,23} - E_{i,0} = 0 \\ 0 \le p_{i,t}^{ch} &\perp \underline{\phi}_{i,t}^{ch} \ge 0, \qquad 0 \le \overline{P}_i^{ch} - p_{i,t}^{ch} \perp \overline{\phi}_{i,t}^{ch} \ge 0 \\ 0 \le p_{i,t}^{dis} &\perp \underline{\phi}_{i,t}^{dis} \ge 0, \qquad 0 \le \overline{P}_i^{dis} - p_{i,t}^{dis} \perp \overline{\phi}_{i,t}^{dis} \ge 0 \\ 0 \le e_{i,t} &\perp \underline{\psi}_{i,t} \ge 0, \qquad 0 \le \overline{E}_i - e_{i,t} \perp \overline{\psi}_{i,t} \ge 0 \end{aligned}$$

#### What Changes from Assignment 1 & Non-Uniqueness of the Best Response

* **No tariff wedge inside the community:** In Assignment 1, $\tau^{imp} + \tau^{exp} > 0$ created a price gap $p_t^{exp} < \lambda_{i,t} < p_t^{imp}$ where a consumer could self-consume without trading. Here, buying and selling happen at the single internal price $x_t$, pinning the shadow price of the member's balance equation to $\lambda_{i,t} = x_t$ in every hour.


* **Flat optimal response sets (Indifference / Degeneracy):** Because the objective of $\mathcal{P}_i(x)$ is completely linear in $(L_{i,t}, PV_{i,t}, p_{i,t}^{ch}, p_{i,t}^{dis})$, a member's optimal response $y_i^*(x)$ is a point only when $x_t$ strictly separates the marginal costs/utilities, and becomes a **convex set (a line segment)** whenever a comparison is a tie:


* *Load tie:* From stationarity, $x_t - \mu_{i,E} = -\mu_{i,t}^{\Delta} + \mu_{i,t}^{-} - \mu_{i,t}^{+}$. Whenever $x_t - \mu_{i,E} = \pm c_i^L$, we have $\mu_{i,t}^{-} = \mu_{i,t}^{+} = 0$, and the member is economically indifferent between any load level $L_{i,t}$ in $[\ell_{i,t}^{ref}, L_{i,t}^{max}]$ (for $-c_i^L$) or $[L_{i,t}^{min}, \ell_{i,t}^{ref}]$ (for $+c_i^L$).


* *PV tie:* Whenever $x_t = c_i^{PV}$, $\nu_{i,t}^{-} = \nu_{i,t}^{+} = 0$, and the member is indifferent to any PV production $PV_{i,t} \in [0, PV_{i,t}^{max}]$.


* *Battery tie:* Whenever $x_t = \eta_i^{ch} \xi_{i,t}$ or $x_t = \xi_{i,t} / \eta_i^{dis}$, the battery owner is indifferent across any charging/discharging rate within $[0, \overline{P}_i^{ch}]$ or $[0, \overline{P}_i^{dis}]$.

### 2. The Community Manager's Bilevel Problem (Model $\mathcal{DP}$)

In the bilevel leader-follower game, the community manager (the leader) moves first by posting the hourly internal prices $x_t$ for all $t \in \mathcal{T}$, anticipating the optimal reactions $y_i \in \arg\min \mathcal{P}_i(x)$ of all 16 community members (the followers).

#### Solution Concept: Optimistic vs. Pessimistic Bilevel Formulation

As shown in Slide 23 of the Week 7 lecture notes, when a follower's problem is degenerate and admits multiple optimal solutions $y_i^*(x)$ at a tie, the leader must specify a tie-breaking rule:

* **Optimistic assumption (standard MPEC):** When both the upper-level variables $x$ and the lower-level primal/dual variables $(y_i, \Phi_i)$ are optimized simultaneously by the leader subject to the followers' KKT conditions, the solver automatically selects the element $y_i \in y_i^*(x)$ that minimizes the leader's objective (best case for the community manager).


* **Pessimistic assumption:** The leader guards against the worst-case optimal response $y_i \in y_i^*(x)$ that a follower could pick when indifferent, which may expose the manager to unexpected cap exceedances $P_t^{exc}$ and penalties $\alpha^{DSO}$ in practice if members break ties uncooperatively.



#### Mathematical Formulation of the Bilevel Problem $\mathcal{DP}$

The manager's goal is to minimize the total community social cost—defined as the sum of the manager's grid payments to the DSO at the shared connection point plus the sum of all members' private PV generation costs and load disutilities (the exact same social welfare objective as model $\mathcal{C}$ in Question 1). Notice that the internal settlement payments $\sum_{i \in \mathcal{I}} x_t q_{i,t}$ are transfers between the members and the manager, so they cancel out in the community's social welfare objective.

The upper-level decision variables of the community manager are the posted internal prices $x_t \in \mathbb{R}$ and the community's metered grid exchanges $P_t^{imp}, P_t^{exp}, P_t^{exc} \ge 0$ for all $t \in \mathcal{T}$, while the lower-level variables are each member's primal schedule $y_i$ and dual multipliers $\Phi_i$:

$$\min_{\substack{x_t, \, P_t^{imp}, \, P_t^{exp}, \, P_t^{exc}, \\ \{y_i, \, \Phi_i\}_{i \in \mathcal{I}}}} \quad \sum_{t=0}^{23} \left( p_t (P_t^{imp} - P_t^{exp}) + (1 - \beta^{DSO})\tau^{imp} P_t^{imp} + (1 - \beta^{DSO})\tau^{exp} P_t^{exp} + (\alpha^{DSO} - (1 - \beta^{DSO})\tau^{imp}) P_t^{exc} + \sum_{i \in \mathcal{I}} \left[ c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right] \right)$$

$$\begin{aligned} \text{s.t.} \quad \sum_{i \in \mathcal{I}} q_{i,t} - P_t^{imp} + P_t^{exp} &= 0 && \forall t \in \mathcal{T} \quad \text{(Community meter balance)} \\ P_t^{imp} - P_t^{exc} - \overline{P}_t^{DSO} &\le 0 && \forall t \in \mathcal{T} \quad \text{(DSO capacity limit)} \\ P_t^{exc} - P_t^{imp} &\le 0 && \forall t \in \mathcal{T} \quad \text{(Exceedance definition)} \\ P_t^{imp} \ge 0, \quad P_t^{exp} \ge 0, \quad P_t^{exc} &\ge 0 && \forall t \in \mathcal{T} \\ (y_i, \Phi_i) \in \text{KKT}\big(\mathcal{P}_i(x)\big) &\quad && \forall i \in \mathcal{I} \quad \text{(Followers' optimal responses)} \end{aligned}$$

(When replacing each follower's problem $\mathcal{P}_i(x)$ with its KKT system $\text{KKT}(\mathcal{P}_i(x))$, the bilevel problem becomes a single-level Mathematical Program with Equilibrium Constraints (MPEC). In Step 3, the non-convex complementarity conditions $0 \le a \perp b \ge 0$ are linearized using Fortuny-Amat big-$M$ binary constraints or Gurobi SOS1 constraints, and in the variants $\mathcal{DP}^x$, the bilinear internal settlement terms $\sum_{t=0}^{23} x_t q_{i,t}$ appearing in the Budget Balance and Individual Rationality constraints are replaced by each member's linear dual objective expression via Strong Duality.)

### 1. Bilevel Formulation of the Manager's Problem ($\mathcal{DP}$ and $\mathcal{DP}^x$)

The daily problem of the community manager (the leader) is to choose the hourly internal prices $x_t$ for the following day to minimize the total social cost of the community, while anticipating the optimal self-scheduling responses of all community members $i \in \mathcal{I}$ (the followers).   Let $C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc})$ denote the hourly metered grid and penalty cost paid by the manager to the DSO at the shared connection point:

   $$C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc}) = p_t (P_t^{imp} - P_t^{exp}) + (1 - \beta^{DSO})\tau^{imp} P_t^{imp} + (1 - \beta^{DSO})\tau^{exp} P_t^{exp} + \big(\alpha^{DSO} - (1 - \beta^{DSO})\tau^{imp}\big) P_t^{exc}$$

and let $C_{i,t}^{\text{priv}}(PV_{i,t}, \Delta_{i,t}^+, \Delta_{i,t}^-) = c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-)$ denote member $i$'s private generation and disutility cost in hour $t$. Because the internal settlement payments $\sum_{i \in \mathcal{I}} \sum_{t=0}^{23} x_t q_{i,t}$ are internal transfers between the members (who pay $x_t q_{i,t}$) and the manager (who collects $x_t q_{i,t}$), they cancel out in the community social cost:

   $$\underbrace{\sum_{t=0}^{23} C_t^{\text{DSO}} - \sum_{i \in \mathcal{I}} \sum_{t=0}^{23} x_t q_{i,t}}_{\text{Manager's net cost}} + \underbrace{\sum_{i \in \mathcal{I}} \sum_{t=0}^{23} \left( x_t q_{i,t} + C_{i,t}^{\text{priv}} \right)}_{\text{Sum of members' total costs}} = \sum_{t=0}^{23} \left( C_t^{\text{DSO}} + \sum_{i \in \mathcal{I}} C_{i,t}^{\text{priv}} \right)$$
   
   In the extended variants $\mathcal{DP}^x$, the manager must also satisfy two economic requirements on the settlement:   Budget Balance (BB): The total internal payments collected from the members over the day must cover (or equal) the manager's costs at the connection point:

   $$\sum_{i \in \mathcal{I}} \sum_{t=0}^{23} x_t q_{i,t} = \sum_{t=0}^{23} C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc}) \quad \text{(or } \ge \text{ for weak BB)}$$
   
   Individual Rationality (IR): No member $i \in \mathcal{I}$ can be worse off signing the community contract than standing alone under their benchmark outside option $C_i^{\text{bench}}$ (either $C_i^{\mathcal{R},*}$ from model $\mathcal{R}$ or $C_i^{\mathcal{I},*}$ from model $\mathcal{I}_i$):
   
   $$\sum_{t=0}^{23} \left( x_t q_{i,t} + c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right) \le C_i^{\text{bench}} \qquad \forall i \in \mathcal{I}$$
   
   The bilevel problem is therefore written as:   
   
   $$\min_{\substack{x_t, \, P_t^{imp}, \, P_t^{exp}, \, P_t^{exc}, \\ \{y_i\}_{i \in \mathcal{I}}}} \quad \sum_{t=0}^{23} \left( C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc}) + \sum_{i \in \mathcal{I}} \left[ c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right] \right)$$
   
   $$\begin{aligned} \text{s.t.} \quad \sum_{i \in \mathcal{I}} q_{i,t} - P_t^{imp} + P_t^{exp} &= 0 && \forall t \in \mathcal{T} \quad \text{(Community meter balance)} \\ P_t^{imp} - P_t^{exc} - \overline{P}_t^{DSO} &\le 0 && \forall t \in \mathcal{T} \quad \text{(DSO capacity limit)} \\ P_t^{exc} - P_t^{imp} &\le 0 && \forall t \in \mathcal{T} \quad \text{(Exceedance definition)} \\ P_t^{imp} \ge 0, \quad P_t^{exp} \ge 0, \quad P_t^{exc} &\ge 0 && \forall t \in \mathcal{T} \\ \text{(Optional BB)} \quad \sum_{t=0}^{23} C_t^{\text{DSO}} - \sum_{i \in \mathcal{I}} \sum_{t=0}^{23} x_t q_{i,t} &= 0 \\ \text{(Optional IR)} \quad \sum_{t=0}^{23} \left( x_t q_{i,t} + C_{i,t}^{\text{priv}} \right) &\le C_i^{\text{bench}} && \forall i \in \mathcal{I} \\ y_i \in \arg\min \mathcal{P}_i(x) &\quad && \forall i \in \mathcal{I} \quad \text{(Followers' best responses)} \end{aligned}$$
   
   ### 2. Step-by-Step Tractable Reformulation into a Single-Level MILP
   
   Replacing each follower's linear program $\mathcal{P}_i(x)$ with its necessary and sufficient KKT conditions turns the bilevel problem into a single-level Mathematical Program with Equilibrium Constraints (MPEC). As shown in Slides 18–22 of the Week 7 lecture notes, this MPEC contains two sources of non-convexity that must be linearized to obtain a Mixed-Integer Linear Program (MILP):   Bilinear terms $x_t q_{i,t}$ (products of the upper-level price $x_t$ and the lower-level net consumption $q_{i,t}$) appearing in the Budget Balance and Individual Rationality constraints.   Bilinear complementarity slackness conditions $0 \le a \perp b \ge 0$ in each follower's KKT system.   Linearization 1: Eliminating the Bilinear Settlement $\sum_{t=0}^{23} x_t q_{i,t}$ via Strong DualityFollowing Slide 19 of Week 7 and the dual derivation in Question 1.c and 3.b of Assignment 1, we derive the dual problem of follower $i$'s linear program $\mathcal{P}_i(x)$ for a given price vector $x$.   Collecting all constant right-hand-side terms in the follower's Lagrangian $\mathcal{L}_i$ (after substituting $\lambda_{i,t} = x_t$ from the stationarity of $q_{i,t}$) yields the dual objective $g_i(\Phi_i)$ of member $i$:
   
   $$g_i(\Phi_i) = \sum_{t=0}^{23} \Big( -\mu_{i,t}^{\Delta} \ell_{i,t}^{ref} + \mu_{i,t}^{-} L_{i,t}^{min} - \mu_{i,t}^{+} L_{i,t}^{max} - \nu_{i,t}^{+} PV_{i,t}^{max} \Big) + \mu_{i,E} E_{i,min} - \mathbb{I}_{i \in \mathcal{I}^{bat}} \left[ E_{i,0}(\xi_{i,0} + \xi_{i,end}) + \sum_{t=0}^{23} \Big( \overline{P}_i^{ch} \overline{\phi}_{i,t}^{ch} + \overline{P}_i^{dis} \overline{\phi}_{i,t}^{dis} + \overline{E}_i \overline{\psi}_{i,t} \Big) \right]$$
   
   By Strong Duality for linear programs (primal objective $=$ dual objective at any KKT point):

   $$\sum_{t=0}^{23} \left( x_t q_{i,t} + c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right) = g_i(\Phi_i) \qquad \forall i \in \mathcal{I}$$
   
   Rearranging this identity expresses member $i$'s daily internal payment $S_i = \sum_{t=0}^{23} x_t q_{i,t}$ as a purely linear function of the follower's primal and dual variables:
   
   $$S_i = \sum_{t=0}^{23} x_t q_{i,t} = g_i(\Phi_i) - \sum_{t=0}^{23} \left( c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right) \qquad \forall i \in \mathcal{I}$$Linearized Budget Balance (BB) constraint:$$\sum_{i \in \mathcal{I}} \left[ g_i(\Phi_i) - \sum_{t=0}^{23} \left( c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right) \right] = \sum_{t=0}^{23} C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc})$$
   
   Linearized Individual Rationality (IR) constraint: Since member $i$'s total cost $\sum_{t=0}^{23} (x_t q_{i,t} + C_{i,t}^{\text{priv}})$ equals $g_i(\Phi_i)$ directly by Strong Duality, the IR constraint is simply:
   
   $$g_i(\Phi_i) \le C_i^{\text{bench}} \qquad \forall i \in \mathcal{I}$$
   
   Linearization 2: Disjunctive (Big-$M$) Reformulation of the Complementarity ConditionsFollowing Slide 21 of Week 7 (Fortuny-Amat & McCarl linearization), each complementarity condition $0 \le a \perp b \ge 0$ is replaced by a binary variable $u \in \{0, 1\}$ and two linear constraints with big-$M$ constants $M^P, M^D > 0$:
   
   $$0 \le a \le M^P (1 - u), \qquad 0 \le b \le M^D u, \qquad u \in \{0, 1\}$$
   
   For each member $i \in \mathcal{I}$ and hour $t \in \mathcal{T}$, this gives:
   
   $$\begin{aligned} 0 \le L_{i,t} - L_{i,t}^{min} &\le (L_{i,t}^{max} - L_{i,t}^{min})(1 - u_{i,t}^{L,-}), & 0 \le \mu_{i,t}^{-} &\le M^{\mu} u_{i,t}^{L,-} \\ 0 \le L_{i,t}^{max} - L_{i,t} &\le (L_{i,t}^{max} - L_{i,t}^{min})(1 - u_{i,t}^{L,+}), & 0 \le \mu_{i,t}^{+} &\le M^{\mu} u_{i,t}^{L,+} \\ 0 \le \sum_{t=0}^{23} L_{i,t} - E_{i,min} &\le M^{E}(1 - u_{i}^{E}), & 0 \le \mu_{i,E} &\le M^{\mu} u_{i}^{E} \\ 0 \le PV_{i,t} &\le PV_{i,t}^{max}(1 - u_{i,t}^{PV,-}), & 0 \le \nu_{i,t}^{-} &\le M^{\nu} u_{i,t}^{PV,-} \\ 0 \le PV_{i,t}^{max} - PV_{i,t} &\le PV_{i,t}^{max}(1 - u_{i,t}^{PV,+}), & 0 \le \nu_{i,t}^{+} &\le M^{\nu} u_{i,t}^{PV,+} \\ 0 \le \Delta_{i,t}^+ &\le \Delta L^{\max}(1 - u_{i,t}^{\Delta,+}), & 0 \le \omega_{i,t}^{+} &\le 2 c_i^L u_{i,t}^{\Delta,+} \\ 0 \le \Delta_{i,t}^- &\le \Delta L^{\max}(1 - u_{i,t}^{\Delta,-}), & 0 \le \omega_{i,t}^{-} &\le 2 c_i^L u_{i,t}^{\Delta,-} \end{aligned}$$

And for battery owners ($i \in \mathcal{I}^{bat}$, $\forall t \in \mathcal{T}$):
  
   $$\begin{aligned} 0 \le p_{i,t}^{ch} &\le \overline{P}_i^{ch}(1 - u_{i,t}^{ch,-}), & 0 \le \underline{\phi}_{i,t}^{ch} &\le M^{\phi} u_{i,t}^{ch,-} \\ 0 \le \overline{P}_i^{ch} - p_{i,t}^{ch} &\le \overline{P}_i^{ch}(1 - u_{i,t}^{ch,+}), & 0 \le \overline{\phi}_{i,t}^{ch} &\le M^{\phi} u_{i,t}^{ch,+} \\ 0 \le p_{i,t}^{dis} &\le \overline{P}_i^{dis}(1 - u_{i,t}^{dis,-}), & 0 \le \underline{\phi}_{i,t}^{dis} &\le M^{\phi} u_{i,t}^{dis,-} \\ 0 \le \overline{P}_i^{dis} - p_{i,t}^{dis} &\le \overline{P}_i^{dis}(1 - u_{i,t}^{dis,+}), & 0 \le \overline{\phi}_{i,t}^{dis} &\le M^{\phi} u_{i,t}^{dis,+} \\ 0 \le e_{i,t} &\le \overline{E}_i(1 - u_{i,t}^{e,-}), & 0 \le \underline{\psi}_{i,t} &\le M^{\psi} u_{i,t}^{e,-} \\ 0 \le \overline{E}_i - e_{i,t} &\le \overline{E}_i(1 - u_{i,t}^{e,+}), & 0 \le \overline{\psi}_{i,t} &\le M^{\psi} u_{i,t}^{e,+} \end{aligned}$$

### 3. Complete Single-Level MILP Formulation of $\mathcal{DP}^x$ (with Slack Variables for Infeasibility)

When a single uniform price $x_t$ cannot simultaneously satisfy strict Budget Balance and strict Individual Rationality for every member, the problem can become infeasible. To handle and quantify this infeasibility within a single parametrized model (as required in Question 2), we introduce non-negative slack variables:   $\sigma^{\text{BB}} \ge 0$: the manager's budget deficit $[\text{DKK}]$ if collected member payments fall short of DSO costs;$\sigma_i^{\text{IR}} \ge 0$: the individual rationality violation $[\text{DKK}]$ for member $i$ (how much member $i$ would have to be compensated to accept the contract), penalized in the objective by a large weight $M_{\text{pen}}$ (or minimized directly to identify the blocking members and the minimum subsidy required).   
The complete single-level MILP is:

$$\min_{\substack{x_t, \, P_t^{imp}, \, P_t^{exp}, \, P_t^{exc}, \, \sigma^{\text{BB}}, \\ \{y_i, \, \Phi_i, \, u_i, \, \sigma_i^{\text{IR}}\}_{i \in \mathcal{I}}}} \quad \sum_{t=0}^{23} \left( C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc}) + \sum_{i \in \mathcal{I}} \left[ c_i^{PV} PV_{i,t} + c_i^L (\Delta_{i,t}^+ + \Delta_{i,t}^-) \right] \right) + M_{\text{pen}} \left( \sigma^{\text{BB}} + \sum_{i \in \mathcal{I}} \sigma_i^{\text{IR}} \right)$$

subject to:Upper-level DSO & community balance constraints ($\forall t \in \mathcal{T}$):

$$\sum_{i \in \mathcal{I}} q_{i,t} - P_t^{imp} + P_t^{exp} = 0, \qquad P_t^{imp} - P_t^{exc} \le \overline{P}_t^{DSO}, \qquad P_t^{exc} \le P_t^{imp}, \qquad P_t^{imp}, P_t^{exp}, P_t^{exc} \ge 0$$Follower primal feasibility & stationarity constraints ($\forall i \in \mathcal{I}, \, \forall t \in \mathcal{T}$):$$\begin{aligned} L_{i,t} - PV_{i,t} + p_{i,t}^{ch} - p_{i,t}^{dis} - q_{i,t} &= 0, & L_{i,t} - \ell_{i,t}^{ref} - \Delta_{i,t}^+ + \Delta_{i,t}^- &= 0 \\ x_t + \mu_{i,t}^{\Delta} - \mu_{i,t}^{-} + \mu_{i,t}^{+} - \mu_{i,E} &= 0, & c_i^{PV} - x_t - \nu_{i,t}^{-} + \nu_{i,t}^{+} &= 0 \\ c_i^L - \mu_{i,t}^{\Delta} - \omega_{i,t}^{+} &= 0, & c_i^L + \mu_{i,t}^{\Delta} - \omega_{i,t}^{-} &= 0 \end{aligned}$$

(plus battery SoC dynamics and battery stationarity equations for $i \in \mathcal{I}^{bat}$).

Follower linearized complementarity constraints ($\forall i \in \mathcal{I}, \, \forall t \in \mathcal{T}$):
All big-$M$ disjunctive constraints (or SOS1 constraints) listed in Linearization 2 above.   
Parameterized settlement constraints (activated by switches for steps $\mathcal{DP}$, $\mathcal{DP}^{\text{BB}}$, $\mathcal{DP}^{\text{BB+IR}}$):
  
   $$\begin{aligned} \text{(Budget Balance)} \quad \sum_{t=0}^{23} C_t^{\text{DSO}}(P_t^{imp}, P_t^{exp}, P_t^{exc}) - \sum_{i \in \mathcal{I}} \left[ g_i(\Phi_i) - \sum_{t=0}^{23} C_{i,t}^{\text{priv}} \right] &\le \sigma^{\text{BB}} \\ \text{(Individual Rationality)} \quad g_i(\Phi_i) - C_i^{\text{bench}} &\le \sigma_i^{\text{IR}} \qquad \forall i \in \mathcal{I} \\ \sigma^{\text{BB}} \ge 0, \quad \sigma_i^{\text{IR}} &\ge 0 \qquad \forall i \in \mathcal{I} \end{aligned}$$

(When strict feasibility is enforced, fix $\sigma^{\text{BB}} = 0$ and $\sigma_i^{\text{IR}} = 0$; when testing which members block feasibility, allow $\sigma_i^{\text{IR}} \ge 0$ and inspect which members have $\sigma_i^{\text{IR}} > 0$ at the optimum.)
